```python
#Stack ; last in first out (LIFO)

push  : O(1)
pop   : O(1)
peek   : O(1)
isempty : O(1)
```

```python
# we implement stack in python using list 
st = []

#stack operations 

| Operation | Python         | Meaning            |
| --------- | -------------- | ------------------ |
| Push      | `append(x)`    | Add element        |
| Pop       | `pop()`        | Remove top element |
| Top/Peek  | `st[-1]`       | See top element    |
| Empty     | `len(st) == 0` | Check empty        |
| Size      | `len(st)`      | Number of elements |

st.clear() # to remove everything 


```

```python
st = []

st.append(10)
st.append(20)
st.append(30)

print("Stack:", st)
print("Top:", st[-1])

x = st.pop()
print("Removed:", x)

print("Stack:", st)
```

```python
#check whether element exists 
if 20 in st:
    print("Found")
```

```python
#queue 
| Operation | Python      | Meaning            |
| --------- | ----------- | ------------------ |
| Enqueue   | `append(x)` | Add at rear        |
| Dequeue   | `popleft()` | Remove from front  |
| Front     | `q[0]`      | See first element  |
| Rear      | `q[-1]`     | See last element   |
| Empty     | `not q`     | Check empty        |
| Size      | `len(q)`    | Number of elements |
```

```python
from collections import deque 

q=deque()
q.append(10)
q.append(20)
q.append(30)

x=q.popleft() # x=10
front : q[0]
rear : q[-1]
size : len(q)
```

```python
from collections import deque

q = deque()

q.append(10)
q.append(20)
q.append(30)

print("Queue:", q)
print("Front:", q[0])
print("Rear:", q[-1])

x = q.popleft()

print("Removed:", x)
print("Queue:", q)
```

```python
######## Priority queue (heap)

import heapq 

h=[]
# lower no means higher prioriy ..i.e min heap
heapq.heappush(h,30)
heapq.heappush(h,20)
heapq.heappush(h,10)

print(h[0]) # smallest no 
x=heapq.heappop(h) # removes top/smallest no 
```

```python
| Function           | Meaning                  |
| ------------------ | ------------------------ |
| `heappush(h, x)`   | Insert `x`               |
| `heappop(h)`       | Remove smallest          |
| `h[0]`             | See smallest             |
| `heapify(h)`       | Convert list → heap      |
| `len(h)`           | Size                     |
| `not h`            | Check empty              |
| `heapreplace(h,x)` | Remove smallest + insert |
| `heappushpop(h,x)` | Insert + remove smallest |
```

```python
Min Heap → store normal values

Max Heap → store negative values
```
